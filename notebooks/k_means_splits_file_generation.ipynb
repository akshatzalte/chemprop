# Splits file generation (K-Means)

The notebook for generating and assigning the clusters (`k_means_clustering_splits_generation.ipynb`) is [here](https://github.com/akshatzalte/chemprop/blob/rigr_home/notebooks/k_means_clustering_splits_generation.ipynb). In this notebook, the cluster assignments are simply used to create the `splits.json` files for unaugmented and augmented cases.

```python
import pandas as pd
import json
import random
```

```python
all_data_set = pd.read_csv("dataset/all_data_set.csv")
all_aug_data_set = pd.read_csv("dataset/all_aug_data_set.csv")
```

```python
all_data_set.columns
```

```python
test_val_clusters = \
[([2, 16], [7, 12]),
 ([2, 16], [6]),
 ([2, 16], [1, 3]),
 ([2, 16], [17, 12]),
 ([2, 16], [4, 8])
  ]
```

```python
all_ids = list(range(18))

splits = []

aug_splits = []


for i, (test, val) in enumerate(test_val_clusters):
    test_val = test+val
    print(test_val)
    train = list(set(all_ids) - set(test_val))
    print(train)
    test_indices = all_data_set.index[all_data_set['cluster_id'].isin(test)].tolist()
    train_indices = all_data_set.index[all_data_set['cluster_id'].isin(train)].tolist()
    val_indices = all_data_set.index[all_data_set['cluster_id'].isin(val)].tolist()
    
    new_split = {
        "test": test_indices,
        "train": train_indices,
        "val": val_indices
    }
    
    # Add the new split to the list
    splits.append(new_split)
    
    selected_test = all_data_set.loc[test_indices].parent_smi.tolist()
    selected_train = all_data_set.loc[train_indices].parent_smi.tolist()
    selected_val = all_data_set.loc[val_indices].parent_smi.tolist()

    aug_test_indices = list(all_aug_data_set[all_aug_data_set["parent_smi"].isin(selected_test)].index)
    aug_train_indices = list(all_aug_data_set[all_aug_data_set["parent_smi"].isin(selected_train)].index)
    aug_val_indices = list(all_aug_data_set[all_aug_data_set["parent_smi"].isin(selected_val)].index)

    new_aug_split = {
            "test": aug_test_indices,
            "train": aug_train_indices,
            "val": aug_val_indices
        }
    aug_splits.append(new_aug_split)

# Save to splits.json
# with open('dataset/splits.json', 'w') as file:
#     json.dump(splits, file, indent=4)

# with open('dataset/aug_splits.json', 'w') as file:
#     json.dump(aug_splits, file, indent=4)
```
